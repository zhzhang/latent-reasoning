# Judge Alt-Test by MMAR category and modality

Scores from `outputs/mmar-judging/judge_accuracy.json` (human triple labels vs
each judge). Primary number is **ρ** (average advantage probability): how often
the judge agrees with the other two labelers at least as well as the left-out
human. **llm** / **hum** are leave-one-out agreement; **ω** is the ε-winning
rate (ε = 0.15).

Every grading format with verdicts is included (gold and no-gold). Empty
formats (currently `neutral_no_gt`) are dropped. Category is the MMAR
**top-level layer** only; modality is the clip source mix.

```python
from __future__ import annotations

import json
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

REPO = Path.cwd()
if not (REPO / "outputs" / "mmar-judging").is_dir():
    REPO = Path("/Users/jordan/tinkering/latent-reasoning")
sys.path.insert(0, str(REPO))

from grader import (  # noqa: E402
    JUDGE_FORMATS,
    grade_mode_title,
    parse_judge_key,
)

sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams.update(
    {
        "figure.dpi": 120,
        "savefig.bbox": "tight",
        "axes.titlesize": 12,
        "axes.labelsize": 11,
    }
)

ACCURACY_PATH = REPO / "outputs" / "mmar-judging" / "judge_accuracy.json"

JUDGE_ORDER = [
    "qwen2.5-omni-7b",
    "phi-4-multimodal",
    "gemma-4-e4b",
    "qwen3-omni-instruct",
    "nemotron-3-nano-omni",
    "qwen3.6-35b-a3b-fp8",
    "claude-sonnet-5",
    "gpt-5.6-luna",
]
JUDGE_LABELS = {
    "qwen2.5-omni-7b": "Qwen2.5-Omni 7B",
    "phi-4-multimodal": "Phi-4 Multimodal",
    "gemma-4-e4b": "Gemma 4 E4B",
    "qwen3-omni-instruct": "Qwen3-Omni Instruct",
    "nemotron-3-nano-omni": "Nemotron 3 Nano Omni",
    "qwen3.6-35b-a3b-fp8": "Qwen3.6 35B A3B",
    "claude-sonnet-5": "Claude Sonnet 5",
    "gpt-5.6-luna": "GPT-5.6 Luna",
}
JUDGE_COLORS = {
    "qwen2.5-omni-7b": "#4C78A8",
    "phi-4-multimodal": "#F58518",
    "gemma-4-e4b": "#54A24B",
    "qwen3-omni-instruct": "#B279A2",
    "nemotron-3-nano-omni": "#E45756",
    "qwen3.6-35b-a3b-fp8": "#72B7B2",
    "claude-sonnet-5": "#9D7553",
    "gpt-5.6-luna": "#F2CF5B",
}
CATEGORY_ORDER = [
    "Signal Layer",
    "Perception Layer",
    "Semantic Layer",
    "Cultural Layer",
]
MODALITY_ORDER = [
    "sound",
    "music",
    "speech",
    "mix-sound-music",
    "mix-sound-speech",
    "mix-music-speech",
    "mix-sound-music-speech",
]
METRIC_COLS = [
    "advantage_prob",
    "loo_agree_judge",
    "loo_agree_human",
    "winning_rate",
    "passed",
    "n",
    "n_missing",
    "n_skipped_lt3",
    "n_label_rows",
    "n_annotators",
]


def pretty_judge(name: str) -> str:
    return JUDGE_LABELS.get(name, name)


def mode_title(name: str) -> str:
    return grade_mode_title(name) or name


def judge_palette(order: list[str]) -> list[str]:
    return [JUDGE_COLORS.get(j, "#BAB0AC") for j in order]


def _stat_row(
    key: str, stats: dict, *, mode: str, slice_kind: str, slice_name: str
) -> dict:
    parsed = parse_judge_key(key)
    row = {
        "mode": mode,
        "slice_kind": slice_kind,
        "slice": slice_name,
        "judge_key": key,
        "model": parsed["model"],
        "gold_tag": parsed["gold_tag"],
    }
    for col in METRIC_COLS:
        row[col] = stats.get(col)
    return row


def load_accuracy(path: Path) -> tuple[dict, pd.DataFrame]:
    payload = json.loads(path.read_text(encoding="utf-8"))
    rows: list[dict] = []
    for mode in payload.get("modes") or list(JUDGE_FORMATS):
        table = payload.get(mode) or {}
        if not isinstance(table, dict):
            continue
        for key, stats in table.items():
            if isinstance(stats, dict):
                rows.append(
                    _stat_row(
                        key, stats, mode=mode, slice_kind="overall", slice_name="overall"
                    )
                )
        for slice_kind, bucket_key in (
            ("category", "by_category"),
            ("modality", "by_modality"),
        ):
            for slice_name, tables in (payload.get(bucket_key) or {}).items():
                slice_table = (tables or {}).get(mode) or {}
                if not isinstance(slice_table, dict):
                    continue
                for key, stats in slice_table.items():
                    if isinstance(stats, dict):
                        rows.append(
                            _stat_row(
                                key,
                                stats,
                                mode=mode,
                                slice_kind=slice_kind,
                                slice_name=str(slice_name),
                            )
                        )
    df = pd.DataFrame(rows)
    present_models = list(dict.fromkeys(df["model"].astype(str)))
    judge_order = [j for j in JUDGE_ORDER if j in present_models] + [
        j for j in present_models if j not in JUDGE_ORDER
    ]
    JUDGE_ORDER[:] = judge_order
    mode_order = [
        m
        for m in (payload.get("modes") or list(JUDGE_FORMATS))
        if m in set(df["mode"])
    ]
    df["model"] = pd.Categorical(df["model"], JUDGE_ORDER, ordered=True)
    df["judge_label"] = df["model"].map(pretty_judge)
    df["mode"] = pd.Categorical(df["mode"], mode_order, ordered=True)
    df["mode_label"] = df["mode"].astype(str).map(mode_title)
    df["slice"] = df["slice"].astype(str)
    return payload, df


def pivot_metric(
    table: pd.DataFrame,
    *,
    index: str,
    columns: str,
    value: str,
    index_order: list[str] | None = None,
    column_order: list[str] | None = None,
) -> pd.DataFrame:
    pivot = table.pivot_table(
        index=index, columns=columns, values=value, aggfunc="mean"
    )
    if index_order:
        pivot = pivot.reindex(
            [i for i in index_order if i in pivot.index]
        )
    if column_order:
        pivot = pivot.reindex(
            columns=[c for c in column_order if c in pivot.columns]
        )
    return pivot


def heatmap(
    ax,
    table: pd.DataFrame,
    value: str,
    *,
    index: str,
    columns: str,
    index_order: list[str],
    column_order: list[str],
    index_labels=None,
    column_labels=None,
    title: str = "",
    fmt: str = ".2f",
    center=None,
    vmin=None,
    vmax=None,
    cmap: str = "RdYlGn",
):
    pivot = pivot_metric(
        table,
        index=index,
        columns=columns,
        value=value,
        index_order=index_order,
        column_order=column_order,
    )
    if index_labels is not None:
        pivot.index = [index_labels(i) for i in pivot.index]
    if column_labels is not None:
        pivot.columns = [column_labels(c) for c in pivot.columns]
    sns.heatmap(
        pivot.astype(float),
        annot=True,
        fmt=fmt,
        cmap=cmap,
        center=center,
        vmin=vmin,
        vmax=vmax,
        ax=ax,
        linewidths=0.4,
        linecolor="white",
        cbar_kws={"shrink": 0.8},
    )
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.set_title(title)
    ax.tick_params(axis="x", rotation=0)


payload, acc = load_accuracy(ACCURACY_PATH)
overall = acc.loc[acc["slice_kind"] == "overall"].copy()
by_cat = acc.loc[acc["slice_kind"] == "category"].copy()
by_mod = acc.loc[acc["slice_kind"] == "modality"].copy()
MODE_ORDER = [
    m
    for m in overall["mode"].cat.categories
    if (overall["mode"].eq(m) & overall["n"].fillna(0).gt(0)).any()
]
overall = overall.loc[overall["mode"].isin(MODE_ORDER)].copy()
by_cat = by_cat.loc[by_cat["mode"].isin(MODE_ORDER)].copy()
by_mod = by_mod.loc[by_mod["mode"].isin(MODE_ORDER)].copy()
by_cat["slice"] = pd.Categorical(by_cat["slice"], CATEGORY_ORDER, ordered=True)
present_mod = [m for m in MODALITY_ORDER if m in set(by_mod["slice"])]
by_mod["slice"] = pd.Categorical(by_mod["slice"], present_mod, ordered=True)

print(f"Loaded {ACCURACY_PATH.relative_to(REPO)}")
print(
    f"{int(payload['n_label_rows']):,} labeled shots · "
    f"{int(payload['n_questions'])} questions · ε={payload['epsilon']}"
)
print("Modes with verdicts:", ", ".join(MODE_ORDER))
print("Judges:", ", ".join(pretty_judge(j) for j in JUDGE_ORDER))
overall.sort_values(["mode", "model"]).head()
```

## Overall (all labeled shots)

ρ is the headline Alt-Test score. Gold formats sit near 0.9–0.98; the no-gold
`free` judges drop into the 0.58–0.76 range. Blank cells are judges that were
not run in that format (API / dedicated text judges skip audio-only recipes).
ω is 0 everywhere here — no judge beats the ε = 0.15 winning-rate bar.

```python
show = overall.sort_values(["mode", "model"]).copy()
show_tbl = show.assign(
    judge=show["judge_label"],
    format=show["mode_label"],
)[
    [
        "format",
        "judge",
        "n",
        "n_missing",
        "advantage_prob",
        "loo_agree_judge",
        "loo_agree_human",
        "winning_rate",
        "passed",
    ]
].rename(
    columns={
        "advantage_prob": "ρ",
        "loo_agree_judge": "llm",
        "loo_agree_human": "hum",
        "winning_rate": "ω",
        "n_missing": "miss",
    }
)
display(
    show_tbl.style.format(
        {
            "ρ": "{:.3f}",
            "llm": "{:.3f}",
            "hum": "{:.3f}",
            "ω": "{:.3f}",
            "n": "{:,.0f}",
            "miss": "{:,.0f}",
        }
    )
    .hide(axis="index")
    .background_gradient(subset=["ρ", "llm"], cmap="YlGn", vmin=0.45, vmax=1.0)
)

fig, axes = plt.subplots(1, 2, figsize=(13.4, 4.8))
heatmap(
    axes[0],
    overall,
    "advantage_prob",
    index="model",
    columns="mode",
    index_order=JUDGE_ORDER,
    column_order=MODE_ORDER,
    index_labels=pretty_judge,
    column_labels=lambda m: str(m).replace("_", "\n"),
    title="ρ  (advantage probability)",
    vmin=0.5,
    vmax=1.0,
    center=0.75,
)
heatmap(
    axes[1],
    overall,
    "loo_agree_judge",
    index="model",
    columns="mode",
    index_order=JUDGE_ORDER,
    column_order=MODE_ORDER,
    index_labels=pretty_judge,
    column_labels=lambda m: str(m).replace("_", "\n"),
    title="llm  (leave-one-out agreement)",
    vmin=0.4,
    vmax=1.0,
    center=0.7,
    cmap="mako",
)
fig.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(12.6, 4.6))
sns.barplot(
    data=overall,
    x="judge_label",
    y="advantage_prob",
    hue="mode",
    order=[pretty_judge(j) for j in JUDGE_ORDER],
    hue_order=MODE_ORDER,
    palette="Set2",
    ax=ax,
)
ax.axhline(0.5, color="#333", ls="--", lw=1, label="ρ = 0.5")
ax.set_ylim(0, 1)
ax.set_xlabel("")
ax.set_ylabel("ρ")
ax.set_title("Alt-Test ρ by judge and grading format")
ax.legend(frameon=False, title="", ncols=2)
ax.tick_params(axis="x", rotation=18)
fig.tight_layout()
plt.show()
```

## By MMAR category (top-level)

Layers coarsest → finest: **Signal → Perception → Semantic → Cultural**.
Each panel is one grading format. `n` in the column header is labeled shots
in that layer (same pool for every judge; missing verdicts are in `miss`).

```python
n_cat = (
    by_cat.groupby("slice", observed=True)["n_label_rows"]
    .max()
    .reindex(CATEGORY_ORDER)
)


def cat_col_label(name: str) -> str:
    n = n_cat.get(name)
    n_s = f"{int(n):,}" if pd.notna(n) else "?"
    short = str(name).replace(" Layer", "")
    return f"{short}\n(n={n_s})"


n_modes = len(MODE_ORDER)
fig, axes = plt.subplots(n_modes, 2, figsize=(12.8, 3.4 * n_modes), squeeze=False)
for row, mode in enumerate(MODE_ORDER):
    sub = by_cat.loc[by_cat["mode"].eq(mode)]
    heatmap(
        axes[row, 0],
        sub,
        "advantage_prob",
        index="model",
        columns="slice",
        index_order=JUDGE_ORDER,
        column_order=CATEGORY_ORDER,
        index_labels=pretty_judge,
        column_labels=cat_col_label,
        title=f"ρ · {mode}",
        vmin=0.45,
        vmax=1.0,
        center=0.75,
    )
    heatmap(
        axes[row, 1],
        sub,
        "loo_agree_judge",
        index="model",
        columns="slice",
        index_order=JUDGE_ORDER,
        column_order=CATEGORY_ORDER,
        index_labels=pretty_judge,
        column_labels=cat_col_label,
        title=f"llm · {mode}",
        vmin=0.35,
        vmax=1.0,
        center=0.7,
        cmap="mako",
    )
fig.suptitle("Alt-Test by MMAR category", y=1.01, fontsize=13)
fig.tight_layout()
plt.show()

# ρ minus that judge's overall ρ in the same format.
delta_cat = by_cat.merge(
    overall[["mode", "model", "advantage_prob"]].rename(
        columns={"advantage_prob": "rho_overall"}
    ),
    on=["mode", "model"],
    how="left",
)
delta_cat["delta_rho"] = delta_cat["advantage_prob"] - delta_cat["rho_overall"]

fig, axes = plt.subplots(
    n_modes, 1, figsize=(10.8, 2.8 * n_modes), squeeze=False
)
for row, mode in enumerate(MODE_ORDER):
    heatmap(
        axes[row, 0],
        delta_cat.loc[delta_cat["mode"].eq(mode)],
        "delta_rho",
        index="model",
        columns="slice",
        index_order=JUDGE_ORDER,
        column_order=CATEGORY_ORDER,
        index_labels=pretty_judge,
        column_labels=cat_col_label,
        title=f"Δρ vs overall · {mode}",
        fmt="+.2f",
        center=0.0,
        vmin=-0.15,
        vmax=0.15,
        cmap="RdBu",
    )
fig.suptitle("Category minus overall ρ (same judge × format)", y=1.01, fontsize=13)
fig.tight_layout()
plt.show()

fig, axes = plt.subplots(n_modes, 1, figsize=(10.6, 3.4 * n_modes), sharey=True)
if n_modes == 1:
    axes = [axes]
for ax, mode in zip(axes, MODE_ORDER):
    sub = by_cat.loc[by_cat["mode"].eq(mode)]
    models = [j for j in JUDGE_ORDER if j in set(sub["model"].astype(str))]
    sns.barplot(
        data=sub,
        x="slice",
        y="advantage_prob",
        hue="judge_label",
        order=CATEGORY_ORDER,
        hue_order=[pretty_judge(j) for j in models],
        palette=judge_palette(models),
        ax=ax,
    )
    ax.axhline(0.5, color="#333", ls="--", lw=1)
    ax.set_ylim(0, 1)
    ax.set_xlabel("")
    ax.set_ylabel("ρ")
    ax.set_title(mode)
    ax.legend(frameon=False, title="", ncols=2, loc="lower right")
fig.suptitle("ρ by MMAR category", y=1.01, fontsize=13)
fig.tight_layout()
plt.show()
```

## By MMAR modality

Single-source clips (`sound` / `music` / `speech`) vs mixtures. `mix-sound-music`
is absent from this labeled pack. Mixture slices are smaller — especially
`mix-sound-music-speech`.

```python
mod_order = [m for m in MODALITY_ORDER if m in set(by_mod["slice"].astype(str))]
n_mod = (
    by_mod.groupby("slice", observed=True)["n_label_rows"]
    .max()
    .reindex(mod_order)
)


def mod_col_label(name: str) -> str:
    n = n_mod.get(name)
    n_s = f"{int(n):,}" if pd.notna(n) else "?"
    return f"{name}\n(n={n_s})"


fig, axes = plt.subplots(n_modes, 2, figsize=(14.2, 3.6 * n_modes), squeeze=False)
for row, mode in enumerate(MODE_ORDER):
    sub = by_mod.loc[by_mod["mode"].eq(mode)]
    heatmap(
        axes[row, 0],
        sub,
        "advantage_prob",
        index="model",
        columns="slice",
        index_order=JUDGE_ORDER,
        column_order=mod_order,
        index_labels=pretty_judge,
        column_labels=mod_col_label,
        title=f"ρ · {mode}",
        vmin=0.40,
        vmax=1.0,
        center=0.75,
    )
    heatmap(
        axes[row, 1],
        sub,
        "loo_agree_judge",
        index="model",
        columns="slice",
        index_order=JUDGE_ORDER,
        column_order=mod_order,
        index_labels=pretty_judge,
        column_labels=mod_col_label,
        title=f"llm · {mode}",
        vmin=0.30,
        vmax=1.0,
        center=0.7,
        cmap="mako",
    )
    axes[row, 0].tick_params(axis="x", rotation=18)
    axes[row, 1].tick_params(axis="x", rotation=18)
fig.suptitle("Alt-Test by MMAR modality", y=1.01, fontsize=13)
fig.tight_layout()
plt.show()

delta_mod = by_mod.merge(
    overall[["mode", "model", "advantage_prob"]].rename(
        columns={"advantage_prob": "rho_overall"}
    ),
    on=["mode", "model"],
    how="left",
)
delta_mod["delta_rho"] = delta_mod["advantage_prob"] - delta_mod["rho_overall"]

fig, axes = plt.subplots(
    n_modes, 1, figsize=(13.2, 2.9 * n_modes), squeeze=False
)
for row, mode in enumerate(MODE_ORDER):
    heatmap(
        axes[row, 0],
        delta_mod.loc[delta_mod["mode"].eq(mode)],
        "delta_rho",
        index="model",
        columns="slice",
        index_order=JUDGE_ORDER,
        column_order=mod_order,
        index_labels=pretty_judge,
        column_labels=mod_col_label,
        title=f"Δρ vs overall · {mode}",
        fmt="+.2f",
        center=0.0,
        vmin=-0.20,
        vmax=0.20,
        cmap="RdBu",
    )
    axes[row, 0].tick_params(axis="x", rotation=18)
fig.suptitle("Modality minus overall ρ (same judge × format)", y=1.01, fontsize=13)
fig.tight_layout()
plt.show()

fig, axes = plt.subplots(n_modes, 1, figsize=(13.0, 3.6 * n_modes), sharey=True)
if n_modes == 1:
    axes = [axes]
for ax, mode in zip(axes, MODE_ORDER):
    sub = by_mod.loc[by_mod["mode"].eq(mode)]
    models = [j for j in JUDGE_ORDER if j in set(sub["model"].astype(str))]
    sns.barplot(
        data=sub,
        x="slice",
        y="advantage_prob",
        hue="judge_label",
        order=mod_order,
        hue_order=[pretty_judge(j) for j in models],
        palette=judge_palette(models),
        ax=ax,
    )
    ax.axhline(0.5, color="#333", ls="--", lw=1)
    ax.set_ylim(0, 1)
    ax.set_xlabel("")
    ax.set_ylabel("ρ")
    ax.set_title(mode)
    ax.legend(frameon=False, title="", ncols=2, loc="lower right")
    ax.tick_params(axis="x", rotation=12)
fig.suptitle("ρ by MMAR modality", y=1.01, fontsize=13)
fig.tight_layout()
plt.show()
```
